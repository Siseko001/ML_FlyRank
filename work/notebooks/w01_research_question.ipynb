## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# CTR by position tier (avg_position=0 means "no data", so exclude it)
visible = df[df.avg_position > 0]
ctr_by_tier = (
    visible.groupby("position_tier")["ctr"]
    .agg(["mean", "median", "count"])
    .sort_values("mean", ascending=False)
)
print("Mean/median CTR (%) by position tier:")
print(ctr_by_tier)

# A high-volume, visible pool to score opportunity within
pool = df[(df.impressions_90d >= 500) & (df.avg_position > 0) & (df.avg_position <= 20)].copy()

# Naive flat-threshold rule (starter baseline's low_ctr_visible_page reason code)
flat_flagged = pool[pool.ctr < 0.5]
print(f"\nFlat CTR<0.5% rule flags {len(flat_flagged)}/{len(pool)} "
      f"({len(flat_flagged) / len(pool):.1%}) of the visible, high-volume pool")

# Position-relative gap: compare each page only to its own tier
pool["tier_median_ctr"] = pool.groupby("position_tier")["ctr"].transform("median")
below_median = pool[pool.ctr < pool.tier_median_ctr]
below_half = pool[pool.ctr < pool.tier_median_ctr * 0.5]

print(f"Below own tier's median CTR: {len(below_median)}/{len(pool)} "
      f"({len(below_median) / len(pool):.1%})")
print(f"Below HALF of own tier's median CTR: {len(below_half)}/{len(pool)} "
      f"({len(below_half) / len(pool):.1%})")
```

**What the numbers show:** CTR drops sharply by position tier (mean CTR ranges from ~2.8% at `top_3` down to ~0.15% at `deep`), so any "low CTR" definition has to be relative to tier, not a flat number. Proof: the starter baseline's flat rule (impressions ≥ 500, position 1–20, CTR < 0.5%) flags about 81% of the visible, high-volume pool — it isn't finding anything, it's just restating that most pages aren't in `top_3`. Comparing each page only to its own tier's median CTR is far more selective: about 49% sit below their tier's median, and about 25% (roughly 3,000 pages) sit below *half* their tier's median — a meaningful, defensible gap rather than noise. That ~25% pool is a credible starting point for a CTR opportunity score, and it's the evidence that a position-adjusted method earns its place over a flat rule.
